**Chains in langchain - Runnable **

```python
prompts.invoke({"language":"Hindi","query":"Katha chalna bhau"})

##as we can invoke both llm and prompts both are runnable. Can we connect both ?

res= llm.invoke(prompts.invoke({"language":"Hindi","query":"Katha chalna bhau"}))
print(res)
```

```python
chains = prompts | llm
```

```python
chains
```

```python
res = chains.invoke({"language":"Hindi","query":"Katha chalna bhau"})
print(res.content)
```

If we want to just print the content output not the entire response

```python
from langchain_core.output_parsers import StrOutputParser
out = StrOutputParser()
```

```python
chains = prompts | llm | out
```

```python

res = chains.invoke({"language":"Hindi","query":"what is langchain"})
print(res)
```

```python
prompts = [
    {"role":"user","content":"Hello my name is sagar"},
    {"role":"assistant","content":"Hello Sagar, what can I do for you?"},
    {"role":"user","content":"what's my name"}
]
res = llm.invoke(prompts)
print(res.content)
```

**Manage history**

```python
history = []
while True:
  query = input("User: ")
  if query.lower() == "exit":
    break
  history.append({"role":"user","content":query})
  res = llm.invoke(history)
  history.append({"role":"assistant","content":res.content})
  print(res.content)
```

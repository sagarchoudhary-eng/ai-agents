```python
from langchain_core.prompts import ChatMessagePromptTemplate
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":'you are a translator. Translate in "{language}"'},
    {"role":"user","content":"{query}"},
])
prompts
```

```python
dynamic_prompts = prompts.format_messages(language="Hindi",query="Chapendi anna")
dynamic_prompts
```

```python
res = llm.invoke(dynamic_prompts)
print(res.content)
```

**Chains in langchain - Runnable **

```python
prompts.invoke({"language":"Hindi","query":"Katha chalna bhau"})

##as we can invoke both llm and prompts both are runnable. Can we connect both ?

res= llm.invoke(prompts.invoke({"language":"Hindi","query":"Katha chalna bhau"}))
print(res)
```

```python
chains = prompts | llm
```

```python
chains
```

```python
res = chains.invoke({"language":"Hindi","query":"Katha chalna bhau"})
print(res.content)
```

If we want to just print the content output not the entire response

```python
from langchain_core.output_parsers import StrOutputParser
out = StrOutputParser()
```

```python
chains = prompts | llm | out
```

```python

res = chains.invoke({"language":"Hindi","query":"what is langchain"})
print(res)
```
